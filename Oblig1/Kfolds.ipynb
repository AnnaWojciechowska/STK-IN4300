```python
from sklearn.model_selection import KFold
```

```python
data = [1, 2, 3, 4, 5, 6, 7, 8, 9]

kf = KFold(n_splits=3)

for train_index, test_index in kf.split(data):
    print("Train:", train_index)
    print("Test: ", test_index)
    print()
```

```python
import string
import numpy as np
numbers = range(1,10)
letters = string.ascii_lowercase[:10]

data = np.array([
    [f"{i}{letter}" for letter in letters]
    for i in numbers
])

kf = KFold(n_splits=5)

for train_index, test_index in kf.split(data):
    print("Train:", data[train_index])
    print("Test: ", data[test_index])
    print()
```
